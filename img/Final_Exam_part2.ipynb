## 研究テーマと選んだ理由

本研究では、「幸福度はどの要因によって、どのようなバランスで支えられているのか」
という問いを設定し、世界幸福度レポートのデータを用いて分析を行う。

幸福度に関する研究では、GDPや経済成長が注目されることが多いが、
必ずしも経済的に豊かな国が最も幸福であるとは限らない。
そこで本研究では、幸福度が高い国と低い国を比較し、
経済・家族・健康・自由・信頼などの構成要素の違いに着目する。

単一の要因ではなく、「幸福度を構成する要素の組み合わせ」に注目することで、
幸福度の本質的な特徴を明らかにすることを目的とする。

※本研究では、分析の正確性を保つため、
幸福度の主要指標に欠損値を含むデータは除外して分析を行った。

```python
# 幸福度分析に使う列
use_columns = [
    "Happiness Rank",
    "Happiness Score",
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom",
    "Trust (Government Corruption)",
    "Generosity",
    "year"
]

# 欠損値を含む行を削除
df = df.dropna(subset=use_columns)
```

#### **分析➀：幸福度上位10か国・下位10か国の抽出**

```python
# 分析対象の年（例：最新年）
latest_year = df["year"].max()

df_latest = df[df["year"] == latest_year]

top10 = df_latest.nsmallest(10, "Happiness Rank")
bottom10 = df_latest.nlargest(10, "Happiness Rank")

top10[["Country", "Happiness Score"]]
```

```python
bottom10[["Country", "Happiness Score"]]
```

#### **分析➁：幸福度構成要素の平均比較**

```python
factors = [
    "Economy (GDP per Capita)",
    "Family",
    "Health (Life Expectancy)",
    "Freedom",
    "Trust (Government Corruption)",
    "Generosity"
]

top_mean = top10[factors].mean()
bottom_mean = bottom10[factors].mean()

comparison_df = pd.DataFrame({
    "Top 10 Countries": top_mean,
    "Bottom 10 Countries": bottom_mean
})

comparison_df
```

#### **可視化➀：構成要素の平均比較（棒グラフ）**

```python
comparison_df.plot(kind="bar")
plt.title("Comparison of Happiness Factors (Top vs Bottom 10 Countries)")
plt.xlabel("Factors")
plt.ylabel("Average Score")
plt.xticks(rotation=45, ha="right")
plt.legend()
plt.tight_layout()
plt.show()
```

#### **分析➂：年次変化（上位国 vs 下位国）**

```python
yearly_top = (
    df.sort_values("Happiness Rank")
      .groupby("year")
      .head(10)
      .groupby("year")["Happiness Score"]
      .mean()
)

yearly_bottom = (
    df.sort_values("Happiness Rank", ascending=False)
      .groupby("year")
      .head(10)
      .groupby("year")["Happiness Score"]
      .mean()
)
```

#### **可視化➁：幸福度スコアの年次推移**

```python
plt.plot(yearly_top.index, yearly_top.values, label="Top 10 Countries")
plt.plot(yearly_bottom.index, yearly_bottom.values, label="Bottom 10 Countries")

plt.title("Yearly Change in Happiness Score")
plt.xlabel("Year")
plt.ylabel("Average Happiness Score")
plt.legend()
plt.grid(True)
plt.show()
```

#### **可視化➂：幸福度スコア分布（ヒストグラム）**

```python
plt.hist(df_latest["Happiness Score"], bins=15)
plt.title("Distribution of Happiness Scores (Latest Year)")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")
plt.show()
```

## 結論

分析の結果、幸福度が高い国と低い国では、幸福度を構成する要素のバランスに
大きな違いがあることが分かった。

幸福度上位国では、経済的要因だけでなく、健康、自由、社会的サポートが
全体的に高い水準で保たれている傾向が見られた。
一方、下位国では特に健康や信頼、家族のサポートが低く、
経済以外の要因が幸福度に大きく影響していることが示唆された。

また、年次変化の分析から、幸福度上位国と下位国の差は長期間にわたって
安定して存在しており、幸福度は短期的な変化ではなく、
社会構造全体の影響を強く受けていると考えられる。

以上より、幸福度は単一の要因ではなく、
複数の要素がバランスよく組み合わさることで高まることが明らかとなった。

```python
geo_path = r"C:\Users\shunk\Downloads\world-countries.json"

with open(geo_path, "r", encoding="utf-8") as f:
    world_geo = json.load(f)
```

```python
world_map = folium.Map(location=[20, 0], zoom_start=2)
```

```python
years = sorted(df["year"].unique())

for y in years:
    year_df = df[df["year"] == y]

    layer = folium.FeatureGroup(name=f"Happiness {y}")

    folium.Choropleth(
        geo_data=world_geo,
        data=year_df,
        columns=["Country", "Happiness Score"],
        key_on="feature.properties.name",
        fill_color="YlGnBu",
        fill_opacity=0.7,
        line_opacity=0.2,
        legend_name=f"Happiness Score ({y})"
    ).add_to(layer)

    # ポップアップ（国名＋スコア）
    for _, row in year_df.iterrows():
        folium.GeoJsonTooltip(
            fields=["name"],
            aliases=[f"{row['Country']} : {row['Happiness Score']:.2f}"]
        )

    layer.add_to(world_map)
```
